1. Configuracion del Entorno y Gestion de Directorios
La base de este proyecto es el orden estructural. Antes de procesar cualquier nota, definimos un ecosistema de carpetas que separa los datos de los archivos MIDI y los datos procesados que son secuencias de texto y los pesos del modelo entrenado. Esta separacion garantiza que el flujo de datos sea unidireccional y facil de depurar.

```python
import os

# Definir carpetas base relativas a la ubicación de esta libreta
folders = ['data/processed_pokemon_music', 'models']

for folder in folders:
    if not os.path.exists(folder):
        os.makedirs(folder, exist_ok=True)
        print(f"✅ Carpeta creada: {folder}")
    else:
        print(f"✔ Carpeta ya existe: {folder}")
```

Durante el entrenamiento, la red intenta predecir la siguiente nota en una secuencia. Al enfrentarse a 74 notas unicas, el modelo aprende no solo que nota sigue, sino que intervalos y armonias definen el estilo de 8 bits. Buscamos un error bajo pero con suficiente flexibilidad para evitar el sobreajuste, permitiendo que la IA sea creativa en lugar de simplemente copiar las pistas originales.
